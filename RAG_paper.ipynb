ဒီ code က llama_index library ကိုသုံးပြီး "metagpt.pdf" ဖိုင်ကို ဖတ်တင်တာပါ။ SimpleDirectoryReader က PDF ဖိုင်ကို ဖတ်ပြီး စာရွက်စာတမ်း (documents) အဖြစ် ဒေတာထဲသို့ ပြောင်းပေးတယ်။ အဓိကလုပ်ဆောင်ချက်က PDF ဖိုင်ထဲက အကြောင်းအရာကို နောက်ထပ် စီမံဆောင်ရွက်ဖို့ (ဥပမာ- မေးခွန်းဖြေတဲ့ system အတွက်) အသင့်ပြင်ဆင်ပေးတာပဲ။

```python
from llama_index.core import SimpleDirectoryReader

# load documents
documents = SimpleDirectoryReader(input_files=["metagpt.pdf"]).load_data()
```

ဒီ code က `llama_index` ရဲ့ `SentenceSplitter` ကိုသုံးပြီး ဖတ်ထားတဲ့ documents တွေကို အပိုင်းသေးတွေ (nodes) အဖြစ် ခွဲထုတ်တယ်။ `chunk_size=1024` ဆိုတာက တစ်ခုချင်းစီရဲ့ အရွယ်အစားကို ၁၀၂၄ လုံးခန့် (characters သို့မဟုတ် tokens) အဖြစ် သတ်မှတ်ထားတာ။ ဒီလိုခွဲထုတ်တာက စာသားကို ပိုမိုစီမံလွယ်ကူအောင် (ဥပမာ- indexing ဒါမှမဟုတ် search လုပ်ဖို့) ပြင်ဆင်ပေးတယ်။

```python
from llama_index.core.node_parser import SentenceSplitter

splitter = SentenceSplitter(chunk_size=1024)
nodes = splitter.get_nodes_from_documents(documents)
```

ဒီ code က `llama_index` ရဲ့ global settings ကို configure လုပ်တာပါ။

- `Settings.llm` ကို `OpenAI` ရဲ့ `gpt-3.5-turbo` model နဲ့ သတ်မှတ်တယ်၊ ဒါက language model အနေနဲ့ မေးခွန်းဖြေတဲ့အခါ ဒီ model ကို သုံးမယ်လို့ ဆိုလိုတယ်။
- `Settings.embed_model` ကို `OpenAIEmbedding` ရဲ့ `text-embedding-ada-002` model နဲ့ သတ်မှတ်တယ်၊ ဒါက text တွေကို vector embeddings အဖြစ်ပြောင်းဖို့ အသုံးပြုမယ့် model ပဲ။

ဒီလို settings လုပ်ထားတာက နောက်ဆက်တွဲ indexing နဲ့ querying လုပ်တဲ့အခါ OpenAI ရဲ့ ဒီ model တွေကို အသုံးပြုဖို့ ပြင်ဆင်တာပါ။

```python
from llama_index.core import Settings
from llama_index.llms.openai import OpenAI
from llama_index.embeddings.openai import OpenAIEmbedding

Settings.llm = OpenAI(model="gpt-3.5-turbo")
Settings.embed_model = OpenAIEmbedding(model="text-embedding-ada-002")
```

ဒီ code က `llama_index` ရဲ့ `SummaryIndex` နဲ့ `VectorStoreIndex` ကို အသုံးပြုပြီး `nodes` တွေကနေ နှစ်မျိုးသော index တွေ ဖန်တီးတယ်။

- `SummaryIndex(nodes)`: ဒါက `nodes` တွေကို အသုံးပြုပြီး စာသားရဲ့ အနှစ်ချုပ် (summary) အတွက် index တစ်ခု ဖန်တီးတယ်။ ဒါက စာရွက်စာတမ်းတွေရဲ့ အကျဉ်းချုပ် ဖော်ပြဖို့ အဆင်ပြေတယ်။
- `VectorStoreIndex(nodes)`: ဒါက `nodes` တွေကို vector embeddings အဖြစ်ပြောင်းပြီး ရှာဖွေမှု (search) နဲ့ similarity-based querying လုပ်ဖို့ index တစ်ခု ဖန်တီးတယ်။

ဒီ index နှစ်ခုကို နောက်ပိုင်း မေးခွန်းဖြေဖို့ ဒါမှမဟုတ် စာသားရှာဖွေဖို့ အသုံးပြုနိုင်တယ်။

```python
from llama_index.core import SummaryIndex, VectorStoreIndex

summary_index = SummaryIndex(nodes)
vector_index = VectorStoreIndex(nodes)
```

ဒီ code က `summary_index` နဲ့ `vector_index` ကနေ query engine နှစ်ခု ဖန်တီးတယ်။

- `summary_query_engine = summary_index.as_query_engine(response_mode="tree_summarize", use_async=True)`: `summary_index` ကို အသုံးပြုပြီး query engine တစ်ခု ဖန်တီးတယ်။ `response_mode="tree_summarize"` က အကျဉ်းချုပ်ဖြေတဲ့အခါ hierarchical (သစ်ပင်ပုံစံ) နည်းနဲ့ ဖြေမယ်။ `use_async=True` က asynchronous နည်းနဲ့ မေးခွန်းဖြေဖို့ သတ်မှတ်တယ်။
- `vector_query_engine = vector_index.as_query_engine()`: `vector_index` ကို အသုံးပြုပြီး query engine တစ်ခု ဖန်တီးတယ်။ ဒါက vector-based search နဲ့ similarity အပေါ်မူတည်ပြီး မေးခွန်းတွေကို ဖြေတယ်။

ဒီ query engine တွေကို နောက်ပိုင်း မေးခွန်းတွေ မေးဖို့ အသုံးပြုမယ်။

```python
summary_query_engine = summary_index.as_query_engine(
    response_mode="tree_summarize",
    use_async=True,
)
vector_query_engine = vector_index.as_query_engine()
```

ဒီ code က `llama_index` ရဲ့ `QueryEngineTool` ကိုသုံးပြီး `summary_query_engine` နဲ့ `vector_query_engine` ကို tools အဖြစ် ဖန်တီးတယ်။

- `summary_tool`: `summary_query_engine` ကို အခြေခံပြီး tool တစ်ခု ဖန်တီးတယ်။ ဒါက MetaGPT နဲ့ ပတ်သက်တဲ့ အကျဉ်းချုပ် မေးခွန်းတွေအတွက် အသုံးဝင်တယ်လို့ description မှာ ဖော်ပြထားတယ်။
- `vector_tool`: `vector_query_engine` ကို အခြေခံပြီး tool တစ်ခု ဖန်တီးတယ်။ MetaGPT paper ထဲက သတ်မှတ်ထားတဲ့ context တွေကို ရယူဖို့ အသုံးဝင်တယ်လို့ ဖော်ပြထားတယ်။

ဒီ tools တွေကို နောက်ပိုင်းမှာ မေးခွန်းတွေကို ရွေးချယ်ဖြေဆိုဖို့ (ဥပမာ- summarization ဒါမှမဟုတ် specific info retrieval) အသုံးပြုနိုင်တယ်။

```python
from llama_index.core.tools import QueryEngineTool


summary_tool = QueryEngineTool.from_defaults(
    query_engine=summary_query_engine,
    description=(
        "Useful for summarization questions related to MetaGPT"
    ),
)

vector_tool = QueryEngineTool.from_defaults(
    query_engine=vector_query_engine,
    description=(
        "Useful for retrieving specific context from the MetaGPT paper."
    ),
)
```

ဒီ code က `llama_index` ရဲ့ `RouterQueryEngine` ကိုသုံးပြီး query engine တစ်ခု ဖန်တီးတယ်။

- `RouterQueryEngine`: ဒါက မေးခွန်းတစ်ခုရဲ့ သဘောအရ ဘယ် tool (`summary_tool` ဒါမှမဟုတ် `vector_tool`) ကို သုံးရမလဲ ဆုံးဖြတ်ပြီး query ကို route လုပ်တယ်။
- `selector=LLMSingleSelector.from_defaults()`: `LLMSingleSelector` က မေးခွန်းကို ခွဲခြမ်းစိတ်ဖြာပြီး သင့်တော်တဲ့ tool တစ်ခုကို ရွေးတယ်။
- `query_engine_tools=[summary_tool, vector_tool]`: `summary_tool` (အကျဉ်းချုပ်အတွက်) နဲ့ `vector_tool` (specific context ရယူဖို့) ကို ရွေးချယ်သုံးမယ့် tools အဖြစ် သတ်မှတ်တယ်။
- `verbose=True`: Query processing လုပ်တဲ့အခါ အသေးစိတ် log ထုတ်ပြမယ်။

ဒီ `query_engine` က မေးခွန်းအမျိုးအစားပေါ်မူတည်ပြီး သင့်တော်တဲ့ tool ကို အလိုအလျောက်ရွေးပြီး ဖြေဆိုမယ်။

```python
from llama_index.core.query_engine.router_query_engine import RouterQueryEngine
from llama_index.core.selectors import LLMSingleSelector


query_engine = RouterQueryEngine(
    selector=LLMSingleSelector.from_defaults(),
    query_engine_tools=[
        summary_tool,
        vector_tool,
    ],
    verbose=True
)
```

```python
response = query_engine.query("What is the summary of the document?")
print(str(response))
```

```python
from llama_index.core import SimpleDirectoryReader
from llama_index.core.node_parser import SentenceSplitter
from llama_index.core import Settings
from llama_index.llms.openai import OpenAI
from llama_index.embeddings.openai import OpenAIEmbedding
from llama_index.core import SummaryIndex, VectorStoreIndex
from llama_index.core.tools import QueryEngineTool
from llama_index.core.query_engine.router_query_engine import RouterQueryEngine
from llama_index.core.selectors import LLMSingleSelector




def get_router_query_engine(file_path: str, llm = None, embed_model = None):
    """Get router query engine."""
    llm = llm or OpenAI(model="gpt-3.5-turbo")
    embed_model = embed_model or OpenAIEmbedding(model="text-embedding-ada-002")

    # load documents
    documents = SimpleDirectoryReader(input_files=[file_path]).load_data()

    splitter = SentenceSplitter(chunk_size=1024)
    nodes = splitter.get_nodes_from_documents(documents)

    summary_index = SummaryIndex(nodes)
    vector_index = VectorStoreIndex(nodes, embed_model=embed_model)

    summary_query_engine = summary_index.as_query_engine(
        response_mode="tree_summarize",
        use_async=True,
        llm=llm
    )
    vector_query_engine = vector_index.as_query_engine(llm=llm)

    summary_tool = QueryEngineTool.from_defaults(
        query_engine=summary_query_engine,
        description=(
            "Useful for summarization questions related to MetaGPT"
        ),
    )

    vector_tool = QueryEngineTool.from_defaults(
        query_engine=vector_query_engine,
        description=(
            "Useful for retrieving specific context from the MetaGPT paper."
        ),
    )

    query_engine = RouterQueryEngine(
        selector=LLMSingleSelector.from_defaults(),
        query_engine_tools=[
            summary_tool,
            vector_tool,
        ],
        verbose=True
    )
    return query_engine
```

```python
query_engine = get_router_query_engine("metagpt.pdf")
```

```python
response = query_engine.query("Tell me about the ablation study results?")
print(str(response))
```
